Choose **Run All** to open the form. The short setting builds three rules; the study setting builds 25. Enter your OpenAI key only when you want to generate a new catalog.

```python
from importlib import metadata
from IPython import get_ipython
from packaging.requirements import Requirement

packages = ['ipywidgets>=8,<9', 'numpy>=1.26,<3', 'pandas>=2.1,<4', 'jinja2>=3.1', 'openai>=1.75,<4', 'scikit-learn>=1.5,<2']
missing = []
for item in packages:
    req = Requirement(item)
    try:
        if metadata.version(req.name) in req.specifier:
            continue
    except metadata.PackageNotFoundError:
        pass
    missing.append(item)
if missing:
    import subprocess, sys
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', *missing])
```
